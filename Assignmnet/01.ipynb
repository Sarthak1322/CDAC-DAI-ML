```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd
import numpy as np
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, log_loss
import matplotlib.pyplot as plt
from tqdm import tqdm

wine_quality = fetch_ucirepo(id=186) 
X = wine_quality.data.features 
y = wine_quality.data.targets 

y = y['quality']

X_train,X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25, stratify=y)
scaler = StandardScaler()
X_trn_scl = scaler.fit_transform(X_train)
X_tst_scl = scaler.transform(X_test)

Cs = np.linspace(0.001, 5, 10)
Gs = np.linspace(0.001, 5, 10)
scores = []
for i in tqdm(range(len(Cs))):
    for g in Gs:
        svm = SVC(kernel='rbf', probability=True, random_state=25, C=Cs[i], gamma=g)
        svm.fit(X_trn_scl, y_train)
        y_pred_prob = svm.predict_proba(X_tst_scl)
        scores.append([Cs[i], g, log_loss(y_test, y_pred_prob)])

df_scores = pd.DataFrame(scores, columns=['C','Gamma', 'score'])
df_scores.sort_values('score', ascending=True)

X.columns

wine_quality.data

X.shape

```

```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

os.chdir("C:/Python/Datasets")
milk = pd.read_csv("milk.csv", index_col=0)
scaler = StandardScaler().set_output(transform='pandas')
df_scaled = scaler.fit_transform(milk)

prcomp = PCA(n_components=2).set_output(transform='pandas')
PC_Data = prcomp.fit_transform(df_scaled)

clust = KMeans(n_clusters=2, random_state=25)
clust.fit(df_scaled)

PC_Data['cluster'] = clust.labels_
PC_Data['cluster'] = PC_Data['cluster'].astype(str)

sns.scatterplot(data=PC_Data, x='pca0', y='pca1', hue='cluster')
plt.xlabel("1st PC")
plt.ylabel("2nd PC")
plt.show()

df_copy = milk.copy()
df_copy['cluster'] = clust.labels_
df_copy['cluster'] = df_copy['cluster'].astype(str)

sns.pairplot(data=df_copy, hue='cluster')
plt.show()





```

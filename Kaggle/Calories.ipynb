```python
import numpy as np
import pandas as pd
import os
from sklearn.linear_model import LinearRegression, ElasticNet

os.chdir("C:/Training/Kaggle/Competitions/Playground_Competitions/CalorieExpenditure")

train = pd.read_csv("train.csv", index_col=0)
train.head()

train.isnull().sum()

test = pd.read_csv("test.csv",  index_col=0)
test.isnull().sum()

X_train, y_train = train.drop('Calories', axis=1), train['Calories']
X_train = pd.get_dummies(X_train, drop_first=True)
X_test = test
X_test = pd.get_dummies(X_test, drop_first=True)

lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)

# Checking any non-negative values in `y_pred`

y_pred[y_pred<0]

y_pred[y_pred<0] = 0

y_pred[y_pred<0]

submit = pd.read_csv("sample_submission.csv")
submit['Calories'] = y_pred
submit.to_csv("sbt_lr.csv", index=False)
```

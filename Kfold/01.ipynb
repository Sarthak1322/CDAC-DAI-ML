```python
import numpy as np
import pandas as pd
import os
from sklearn.model_selection import KFold, cross_val_predict, cross_val_score
from sklearn.linear_model import ElasticNet
from sklearn.metrics import r2_score

os.chdir("C:/Python/Datasets")

pizza = pd.read_csv("pizza.csv")
X, y = pizza[['Promote']], pizza['Sales']

el = ElasticNet()

kfold = KFold(n_splits=5, shuffle=True, random_state=25)
predictions = cross_val_predict(el, X, y, cv=kfold)
predictions

results = cross_val_score(el, X, y, cv=kfold)
results

results.mean()

results = cross_val_score(el, X, y, cv=kfold, scoring='neg_root_mean_squared_log_error')
results

results.mean()

kfold = KFold(n_splits=5, shuffle=True, random_state=25)
train_index, test_index = next(iter(kfold.split(pizza)))
train_index, test_index

X_train, y_train = X.iloc[train_index], y.iloc[train_index]
X_test, y_test = X.iloc[test_index], y.iloc[test_index]
X_train.shape, y_train.shape, X_test.shape, y_test.shape

el.fit(X_train, y_train)
y_pred = el.predict(X_test)
print(r2_score(y_test, y_pred))

scores = []
for i, (train_index, test_index) in enumerate(kfold.split(pizza)):
    print(f"Fold {i}:")
    print(f"  Train: index={train_index}")
    X_train = X.iloc[train_index]
    y_train = y.iloc[train_index]
    el.fit(X_train, y_train)

    print(f"  Test:  index={test_index}")    
    X_test = X.iloc[test_index]
    y_test = y.iloc[test_index]
    y_pred = el.predict(X_test)
    print("Predictions:",y_pred)
    
    print(r2_score(y_true=y_test, y_pred=y_pred))
    scores.append(r2_score(y_true=y_test, y_pred=y_pred))


```

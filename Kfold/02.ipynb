```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
import os

os.chdir("C:/Python/Cases/Concrete Strength")

concrete = pd.read_csv("Concrete_Data.csv")
X, y = concrete.drop('Strength', axis=1), concrete['Strength']

rf = RandomForestRegressor(random_state=25)

kfold = KFold(n_splits=5, shuffle=True, random_state=25)
params = {'max_features':[3,4,5,6]}
gcv = GridSearchCV(rf, param_grid=params, cv=kfold, n_jobs=-1)
gcv.fit(X, y)

gcv.best_params_, gcv.best_score_

#### Inferencing

tst = pd.read_csv("testConcrete.csv")

rf = RandomForestRegressor(random_state=25, max_features=4)
rf.fit(X,y)

rf.predict(tst)

# OR

best_model = gcv.best_estimator_
best_model.predict(tst)

# OR

gcv.predict(tst)
```

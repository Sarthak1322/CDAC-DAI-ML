```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from xgboost import XGBRegressor
import matplotlib.pyplot as plt
from sklearn.linear_model import ElasticNet, LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import StackingRegressor
from tqdm import tqdm

concrete = pd.read_csv("C:/Python/Cases/Concrete Strength/Concrete_Data.csv")
X, y = concrete.drop('Strength', axis=1), concrete['Strength']

X_train,X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)

# w/o passthrough

lr = LinearRegression()
el = ElasticNet()
dtr = DecisionTreeRegressor(random_state=25)
gbm = XGBRegressor(random_state=25)
stack = StackingRegressor([('LR',lr),('EL',el),('TREE',dtr)], final_estimator=gbm)

stack.fit(X_train, y_train)
y_pred = stack.predict(X_test)
r2_score(y_test, y_pred)

# with passthrough

stack = StackingRegressor([('LR',lr),('EL',el),('TREE',dtr)], final_estimator=gbm, passthrough=True)
stack.fit(X_train, y_train)
y_pred = stack.predict(X_test)
r2_score(y_test, y_pred)
```

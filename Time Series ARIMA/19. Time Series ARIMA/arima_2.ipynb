```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.metrics import root_mean_squared_error
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.arima.model import ARIMA
from tqdm import tqdm
from pmdarima.arima import auto_arima
```

```python
os.chdir("C:/Python/Datasets")
df = pd.read_csv("FMAC-HPI_24420.csv")
y = df['NSA Value']
```

```python
y_train = y.iloc[:-12]
y_test = y.iloc[-12:]
y.shape, y_train.shape, y_test.shape
```

```python
model = ARIMA(y_train,order=(1,1,1), seasonal_order=(1,1,1,12))
model_fit = model.fit()
y_pred = model_fit.predict(start=len(y_train), end=len(y_train)+len(y_test)-1)
root_mean_squared_error(y_test, y_pred)
```

### `auto_arima`

```python
import warnings
warnings.filterwarnings('ignore')
```

##### ARIMA

```python
model = auto_arima(y_train, trace=True, error_action='ignore', suppress_warnings=True)
```

```python
y_pred = model.predict(n_periods=len(y_test))
root_mean_squared_error(y_test, y_pred)
```

##### SARIMA

```python
s_model = auto_arima(y_train, trace=True,  error_action='ignore', suppress_warnings=True,
                   seasonal=True,m=12)
```

```python
y_pred = s_model.predict(n_periods=len(y_test))
root_mean_squared_error(y_test, y_pred)
```

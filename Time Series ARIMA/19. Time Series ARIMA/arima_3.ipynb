```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.metrics import root_mean_squared_error
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.api import ExponentialSmoothing
from tqdm import tqdm
from pmdarima.arima import auto_arima
```

```python
os.chdir("C:/Training/Kaggle/Competitions/Rossmann Store Sales")
```

```python
train = pd.read_csv("train.csv", low_memory=False)
train['Store'].value_counts()
```

```python
store_1_sales = train[train['Store']==1][['Date','Sales']].reset_index(drop=True)
store_1_sales.shape
```

```python
store_1_sales
```

```python
store_1_sales = store_1_sales.sort_values('Date')
```

```python
y = store_1_sales['Sales']
y_train, y_test = y.iloc[:-35], y.iloc[-35:]
y_train.shape, y_test.shape
```

```python
model = auto_arima(y_train, trace=True, error_action='ignore', suppress_warnings=True)
```

```python
y_pred = model.predict(n_periods=len(y_test))
root_mean_squared_error(y_test, y_pred)
```

```python
y_pred.index = y_test.index
```

```python
s_model = auto_arima(y_train, trace=True,  error_action='ignore', suppress_warnings=True,
                   seasonal=True,m=7)
```

```python
y_pred = s_model.predict(n_periods=len(y_test))
root_mean_squared_error(y_test, y_pred)
```

##### Holt-Winters

```python
hw_add = ExponentialSmoothing(y_train, seasonal_periods=7,  
                              trend='add', seasonal='add')
fit1 = hw_add.fit()
y_pred = fit1.forecast(len(y_test))
root_mean_squared_error(y_test, y_pred)
```

```python
hw_add = ExponentialSmoothing(y_train, seasonal_periods=7,  
                              trend='add', seasonal='add', damped_trend=True)
fit1 = hw_add.fit()
y_pred = fit1.forecast(len(y_test))
root_mean_squared_error(y_test, y_pred)
```

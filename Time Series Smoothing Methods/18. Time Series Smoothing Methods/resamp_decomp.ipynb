```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
os.chdir("C:/Python/Datasets")
```

```python
df = pd.read_csv("monthly-milk-production-pounds-p.csv", index_col=0)
df.index = pd.DatetimeIndex( df.index )
```

```python
df.plot()
plt.title("Monthly Milk Production")
plt.xlabel("Months")
plt.show()
```

```python
downsampled  = df.resample('QE').sum() 
downsampled.index.rename('Quarter', inplace=True)
downsampled.plot()
plt.title("Quarterly Milk Production")
plt.xlabel("Quarters")
plt.show()
```

### Decomposition

```python
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='additive',period=12)
result.plot()
plt.title("Additive Decomposition")
plt.show()
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='multiplicative',period=12)
result.plot()
plt.title("Multiplicative Decomposition")
plt.show()
```

```python
series = downsampled['Milk']
result = seasonal_decompose(series, model='additive',period=4)
result.plot()
plt.title("Additive Decomposition for Quarterly")
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, ElasticNet
from sklearn.preprocessing import  StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import VotingRegressor 
from sklearn.tree import DecisionTreeRegressor

df = pd.read_csv(r"/home/sarthakredasani/Documents/CDAC_ML/Cases/Cases/Concrete Strength/Concrete_Data.csv")
X, y = df.drop('Strength', axis=1), df['Strength']
X_train,X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)

lr = LinearRegression()
el = ElasticNet()
knn = KNeighborsRegressor()
dtr = DecisionTreeRegressor(random_state=25)
voting = VotingRegressor([('LR',lr), ('EL',el), ('DTR',dtr),('KNN',knn)])

voting.fit(X_train, y_train)
y_pred = voting.predict(X_test)
print(r2_score(y_test, y_pred))

# Getting Individual Performances of estimators

voting.named_estimators_

y_pred = voting.named_estimators_['LR'].predict(X_test)
print(r2_score(y_test, y_pred))

y_pred = voting.named_estimators_['EL'].predict(X_test)
print(r2_score(y_test, y_pred))

y_pred = voting.named_estimators_['DTR'].predict(X_test)
print(r2_score(y_test, y_pred))

y_pred = voting.named_estimators_['KNN'].predict(X_test)
print(r2_score(y_test, y_pred))

voting = VotingRegressor([('LR',lr), ('EL',el), ('DTR',dtr),('KNN',knn)],
                         weights=[2,2,6,4])
voting.fit(X_train, y_train)
y_pred = voting.predict(X_test)
print(r2_score(y_test, y_pred))

voting = VotingRegressor([('LR',lr), ('EL',el), ('DTR',dtr),('KNN',knn)],
            weights=[0.6351839142464111,0.6345321364921961,0.8127760533837747,0.7160443418569369])
voting.fit(X_train, y_train)
y_pred = voting.predict(X_test)
print(r2_score(y_test, y_pred))
```
